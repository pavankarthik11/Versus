```python
from collections import defaultdict

player_stats = defaultdict(lambda: {
    "runs": 0,
    "balls_faced": 0,
    "fours": 0,
    "sixes": 0,
    "runs_conceded": 0,
    "balls_bowled": 0,
    "wickets": 0,
    "dismissals": 0
})
```

```python
for over in innings["overs"]:
    for delivery in over["deliveries"]:

        batter = delivery["batter"]
        bowler = delivery["bowler"]

        runs = delivery["runs"]

        # Batter runs
        player_stats[batter]["runs"] += runs["batter"]

        # Count 4s and 6s
        if runs["batter"] == 4:
            player_stats[batter]["fours"] += 1

        if runs["batter"] == 6:
            player_stats[batter]["sixes"] += 1

        # Batter faced a legal ball
        if "extras" not in delivery or (
            "wides" not in delivery["extras"]
            and "noballs" not in delivery["extras"]
        ):
            player_stats[batter]["balls_faced"] += 1

        # Bowler information
        player_stats[bowler]["runs_conceded"] += runs["total"]

        # Bowler faced a legal delivery
        if "extras" not in delivery or (
            "wides" not in delivery["extras"]
            and "noballs" not in delivery["extras"]
        ):
            player_stats[bowler]["balls_bowled"] += 1

        # Wickets
        if "wickets" in delivery:
            for wicket in delivery["wickets"]:

                player_out = wicket["player_out"]

                player_stats[player_out]["dismissals"] += 1

                # Bowler gets credit for most dismissals,
                # but NOT run out / retired hurt / obstructing the field
                if wicket["kind"] not in [
                    "run out",
                    "retired hurt",
                    "retired out",
                    "obstructing the field"
                ]:
                    player_stats[bowler]["wickets"] += 1
```

```python
stats_df = pd.DataFrame.from_dict(
    player_stats,
    orient="index"
)

stats_df
```

```python
stats_df["strike_rate"] = (
    stats_df["runs"] / stats_df["balls_faced"] * 100
).fillna(0)

stats_df["economy"] = (
    stats_df["runs_conceded"] / stats_df["balls_bowled"] * 6
).fillna(0)

stats_df
```

```python
from collections import defaultdict

def calculate_player_stats(innings):

    player_stats = defaultdict(lambda: {
        "runs": 0,
        "balls_faced": 0,
        "fours": 0,
        "sixes": 0,
        "runs_conceded": 0,
        "balls_bowled": 0,
        "wickets": 0,
        "dismissals": 0
    })

    for over in innings["overs"]:
        for delivery in over["deliveries"]:

            batter = delivery["batter"]
            bowler = delivery["bowler"]
            runs = delivery["runs"]

            # -------------------------
            # BATTING
            # -------------------------

            # Batter runs
            player_stats[batter]["runs"] += runs["batter"]

            # Legal ball faced
            if "extras" not in delivery or (
                "wides" not in delivery["extras"]
                and "noballs" not in delivery["extras"]
            ):
                player_stats[batter]["balls_faced"] += 1

            # Fours
            if runs["batter"] == 4:
                player_stats[batter]["fours"] += 1

            # Sixes
            if runs["batter"] == 6:
                player_stats[batter]["sixes"] += 1

            # -------------------------
            # BOWLING
            # -------------------------

            # Legal ball bowled
            if "extras" not in delivery or (
                "wides" not in delivery["extras"]
                and "noballs" not in delivery["extras"]
            ):
                player_stats[bowler]["balls_bowled"] += 1

            # Runs conceded by bowler
            bowler_runs = runs["total"]

            if "extras" in delivery:
                bowler_runs -= delivery["extras"].get("byes", 0)
                bowler_runs -= delivery["extras"].get("legbyes", 0)

            player_stats[bowler]["runs_conceded"] += bowler_runs

            # -------------------------
            # WICKETS
            # -------------------------

            if "wickets" in delivery:

                for wicket in delivery["wickets"]:

                    player_out = wicket["player_out"]
                    wicket_kind = wicket["kind"]

                    # Batter was dismissed
                    player_stats[player_out]["dismissals"] += 1

                    # Bowler gets credit for these wickets
                    if wicket_kind not in [
                        "run out",
                        "retired hurt",
                        "retired out",
                        "obstructing the field"
                    ]:
                        player_stats[bowler]["wickets"] += 1

    return player_stats
```
